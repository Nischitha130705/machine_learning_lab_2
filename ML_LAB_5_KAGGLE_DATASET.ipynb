```python
import pandas as pd
import wittgenstein as lw
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("mushrooms.csv")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nClass distribution:")
print(df["class"].value_counts())
```

```python
X = df.drop("class", axis=1)
y = df["class"]

print("Features:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
model = lw.RIPPER()
```

```python
model.fit(
    df,
    class_feat="class",
    pos_class="e"
)

print("RIPPER model trained successfully!")
```

```python
print("Generated RIPPER Classification Rules")
print("=" * 50)

print(model.out_model())
```

```python
predictions = model.predict(df)

print("First 20 predictions:")
print(predictions[:20])
```

```python
# Map true labels to booleans to match prediction format
y_true_bool = y == 'e'

accuracy = accuracy_score(y_true_bool, predictions)
print("RIPPER Accuracy:", accuracy * 100, "%")
```

```python
new_mushroom = X.iloc[[0]]

print("New Mushroom:")
display(new_mushroom)

prediction = model.predict(new_mushroom)

# prediction is a list of booleans, e.g. [False]
pred_val = prediction[0]
print("Predicted as positive class (edible):", pred_val)

if pred_val:
    print("Prediction: EDIBLE")
else:
    print("Prediction: POISONOUS")
```
