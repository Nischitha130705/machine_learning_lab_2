```python
# ============================================================
# FIND-S ALGORITHM
# Breast Cancer Wisconsin Dataset
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

data = pd.read_csv('/content/data.csv')

print("=" * 60)
print("BREAST CANCER WISCONSIN DATASET")
print("=" * 60)

print("\nDataset Shape:", data.shape)
print("\nFirst 5 Rows:")
print(data.head())


# ------------------------------------------------------------
# 2. DATA PREPROCESSING
# ------------------------------------------------------------

# Remove unnecessary columns
data = data.drop(['id', 'Unnamed: 32'], axis=1)

# Rename diagnosis as target
data = data.rename(columns={'diagnosis': 'target'})

# Convert:
# M = Malignant = 1 (Positive)
# B = Benign    = 0 (Negative)

data['target'] = data['target'].map({
    'M': 1,
    'B': 0
})

print("\n" + "=" * 60)
print("AFTER PREPROCESSING")
print("=" * 60)

print("\nDataset Shape:", data.shape)

print("\nClass Distribution:")
print(data['target'].value_counts())

print("\n1 = Malignant")
print("0 = Benign")


# ------------------------------------------------------------
# 3. PREPARE ATTRIBUTES
# ------------------------------------------------------------

attributes = list(data.columns.drop('target'))

print("\nNumber of Attributes:", len(attributes))


# ------------------------------------------------------------
# 4. CONVERT NUMERICAL DATA TO CATEGORICAL
# ------------------------------------------------------------

# Find-S works better with categorical attributes.
# Each value is converted into:
# High = value >= median
# Low  = value < median

finds_data = data.copy()

for column in attributes:
    median = finds_data[column].median()

    finds_data[column] = finds_data[column].apply(
        lambda x: 'High' if x >= median else 'Low'
    )


print("\n" + "=" * 60)
print("CONVERTED DATA")
print("=" * 60)

print(finds_data.head())


# ------------------------------------------------------------
# 5. INITIALIZE HYPOTHESIS
# ------------------------------------------------------------

hypothesis = ['∅'] * len(attributes)

print("\n" + "=" * 60)
print("INITIAL HYPOTHESIS")
print("=" * 60)

print(hypothesis)


# ------------------------------------------------------------
# 6. FIND-S ALGORITHM
# ------------------------------------------------------------

positive_count = 0
first_positive = True

for _, row in finds_data.iterrows():

    # Consider only positive examples
    # Positive = Malignant = 1

    if row['target'] == 1:

        positive_count += 1

        # First positive example
        if first_positive:

            hypothesis = [
                row[attribute]
                for attribute in attributes
            ]

            first_positive = False

        # Compare with subsequent positive examples
        else:

            for i, attribute in enumerate(attributes):

                if hypothesis[i] != row[attribute]:
                    hypothesis[i] = '?'


# ------------------------------------------------------------
# 7. DISPLAY RESULT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FIND-S RESULT")
print("=" * 60)

print("\nPositive Examples:", positive_count)

print("\nFinal Hypothesis:")
print(hypothesis)


# ------------------------------------------------------------
# 8. ATTRIBUTE-WISE HYPOTHESIS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ATTRIBUTE-WISE FINAL HYPOTHESIS")
print("=" * 60)

for attribute, value in zip(attributes, hypothesis):
    print(f"{attribute:30} : {value}")


# ------------------------------------------------------------
# 9. SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)

print("Dataset              : Breast Cancer Wisconsin")
print("Total Records        :", len(data))
print("Total Attributes     :", len(attributes))
print("Positive Class       : Malignant (M)")
print("Negative Class       : Benign (B)")
print("Positive Examples    :", positive_count)

print("\nFinal Hypothesis:")
print(hypothesis)

print("\nFind-S Algorithm Completed Successfully!")
```

```python
import pandas as pd

# Direct dataset URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/car/car.data"

# Column names
columns = [
    "buying",
    "maint",
    "doors",
    "persons",
    "lug_boot",
    "safety",
    "class"
]

# Load dataset
df = pd.read_csv(url, header=None, names=columns)

print("Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nClass Distribution:")
print(df["class"].value_counts())

# Attributes
attributes = columns[:-1]

# Initialize hypothesis
hypothesis = ["∅"] * len(attributes)

# Find-S Algorithm
for index, row in df.iterrows():

    # Consider positive examples
    # Here, "acc" is considered the positive class
    if row["class"] == "acc":

        # First positive example
        if hypothesis == ["∅"] * len(attributes):
            hypothesis = list(row[attributes])

        else:
            # Compare attributes
            for i in range(len(attributes)):

                if hypothesis[i] != row[attributes[i]]:
                    hypothesis[i] = "?"

print("\nFinal Hypothesis:")
print(hypothesis)
```
